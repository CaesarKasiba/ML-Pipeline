## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*


A simple fixed rule could identify pages for review using a single signal, such as selecting every page with a downward trend. However, this would produce a large group of candidates and would treat pages with very different characteristics in the same way.

A learned ranking approach could combine multiple signals such as impressions, CTR, average position, content age, engagement, content type, and search demand. This may allow the system to prioritize pages more selectively than a single threshold or rule.

ML is therefore useful here not because a model is automatically better than a rule, but because the decision involves multiple interacting signals and the team needs a prioritized list rather than a simple yes/no rule. The model would still need to demonstrate improvement over a sensible baseline before it would be considered useful.
